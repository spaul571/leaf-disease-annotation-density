```python
import os

# Path to your main dataset folder
dataset_path = r"F:\0Defense\Dataset\Vegetables_Dataset_seperate_folder"

# Supported image extensions
image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".tiff", ".webp"}

grand_total = 0  # total across all main folders

# Loop through each main folder (e.g., Tomato, Potato, etc.)
for main_folder in os.listdir(dataset_path):
    main_path = os.path.join(dataset_path, main_folder)
    if not os.path.isdir(main_path):
        continue

    folder_total = 0  # total for this main folder

    # Walk through all subfolders in this main folder
    for root, dirs, files in os.walk(main_path):
        count = sum(
            1 for file in files
            if os.path.splitext(file)[1].lower() in image_extensions
        )
        if count > 0:
            print(f"{root}: {count} images")
            folder_total += count

    print(f"➡️ Total in {main_folder}: {folder_total} images\n")
    grand_total += folder_total

# Print grand total
print("=" * 50)
print(f"🧾 Grand Total: {grand_total} images")
print("=" * 50)
```

```python
import os

# Path to your main dataset folder
base_path = r"F:\0Defense\Dataset\Vegetables_Dataset_seperate_folder_increase_size"

# Supported image extensions
image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".tiff", ".webp"}

# Rename images and labels consistently
def rename_images_and_labels(image_dir, label_dir, prefix):
    images = sorted([
        f for f in os.listdir(image_dir)
        if os.path.splitext(f)[1].lower() in image_extensions
    ])

    for i, img_file in enumerate(images, start=1):
        base_name, ext = os.path.splitext(img_file)
        new_base = f"{prefix}_{i:04d}"  # e.g., Tomato_0001

        old_img_path = os.path.join(image_dir, img_file)
        new_img_path = os.path.join(image_dir, new_base + ext)

        # Rename image
        os.rename(old_img_path, new_img_path)

        # Corresponding label
        old_label_path = os.path.join(label_dir, base_name + ".txt")
        new_label_path = os.path.join(label_dir, new_base + ".txt")

        if os.path.exists(old_label_path):
            os.rename(old_label_path, new_label_path)

# Loop through each vegetable folder
for veg_folder in os.listdir(base_path):
    veg_path = os.path.join(base_path, veg_folder)
    if not os.path.isdir(veg_path):
        continue

    print(f"\n🔄 Renaming files in: {veg_folder}")

    # Process train/test/valid splits
    for split in ["train", "test", "valid"]:
        img_dir = os.path.join(veg_path, split, "images")
        lbl_dir = os.path.join(veg_path, split, "labels")

        if not os.path.exists(img_dir) or not os.path.exists(lbl_dir):
            continue

        rename_images_and_labels(img_dir, lbl_dir, prefix=f"{veg_folder}_{split}")

print("\n✅ Done! All images and labels renamed consistently.")
```
